```python
import numpy as np
```

```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
# Load Dataset
df = pd.read_csv("ipl-matches.csv")
print("First 5 Rows:\n", df.head(10))


# 🏆 Most Successful Teams
team_wins=df['WinningTeam'].value_counts()
print("\nMost Wins by Teams:\n", team_wins)

# 🥇 Most Player of the Match Awards
top_players = df['Player_of_Match'].value_counts().head(10)
print("\nTop Player of the Match Winners:\n", top_players)

# 🏟️ Venues with Most Matches
top_venues = df['Venue'].value_counts().head(5)
print("\nTop Venues:\n", top_venues)

# 📅 Matches Per Season
matches_per_season = df['Season'].value_counts().sort_index()
print("\nMatches Per Season:\n", matches_per_season)



# 📊 Visualization
# Bar chart - Team wins
team_wins.plot(kind='bar', color='seagreen', title="Matches Won by Teams")
plt.xlabel("Teams")
plt.ylabel("Matches Won")
plt.tight_layout()
plt.show()

# Bar chart - Top 10 players
top_players.plot(kind='bar', color='orange', title="Top 10 Players of the Match")
plt.xlabel("Player")
plt.ylabel("Awards")
plt.tight_layout()
plt.show()

# Pie chart - Matches by Venue
top_venues.plot(kind='pie', autopct='%1.1f%%', title="Top 5 Venues (Matches Hosted)")
plt.ylabel("")
plt.tight_layout()
plt.show()

# Line chart - Matches per Season
matches_per_season.plot(kind='line', marker='o', title="Matches Played per IPL Season")
plt.xlabel("Season")
plt.ylabel("Number of Matches")
plt.grid(True)
plt.tight_layout()
plt.show()
```
